```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

main_path = '/Users/Kai/HPL-LATTICeA/'
import os
import sys
sys.path.append(main_path)

from libraries.data_processing import load_topography
```

```python
pure_cores_counts = pd.read_csv(os.path.join(main_path, 'data/6-plex/pure_cores_densities_with_annotations.csv'), index_col=0)
pure_cores_counts
```

```python
pure_cores_counts['cluster'].value_counts()
```

```python
leiden_clusters, malignant_clusters, stromal_clusters = load_topography(main_path=main_path)
all_abnormal = sorted(list(malignant_clusters) + list(stromal_clusters))
```

```python
pure_cores_counts['cluster_'] = pure_cores_counts['cluster'].apply(lambda x: int(x.split(' ')[1]))
pure_cores_counts = pure_cores_counts[pure_cores_counts['cluster_'].isin(all_abnormal)]
```

```python
pure_cores_counts['annotation'] = pure_cores_counts['cluster_'].apply(lambda x: 'Malignant' if x in malignant_clusters else 'Stromal')
```

```python
[c for c in malignant_clusters if c not in pure_cores_counts['cluster_'].to_list()]
```

```python
dec_order_all = list(pure_cores_counts['cluster'].value_counts().to_frame().sort_values(by='count', ascending=False).index)

malignant_colour = "steelblue"
stromal_colour = "#coral"

cluster_palette = {
    'Malignant':malignant_colour,
    'Stromal':stromal_colour
}

sns.set_context("paper", font_scale=2.0)
fig, ax = plt.subplots(figsize=(14,5))

sns.countplot(pure_cores_counts, x='cluster', order=dec_order_all, hue='annotation', ax=ax)
ax.set_xticklabels([x.get_text().split(' ')[1] for x in ax.get_xticklabels()], rotation=90)
ax.set_ylabel('Number of pure cores\nin LATTICeA-6plex')
ax.set_xlabel('Cluster')
ax.legend_.set_title('Annotation')
plt.savefig(os.path.join(main_path, 'analysis/multiplex-if/6-plex/cluster_frequency_6plex_all_abnormal.pdf'), dpi=300, bbox_inches='tight', pad_inches=0.1)
```
