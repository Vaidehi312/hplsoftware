```python
from scipy.stats import median_abs_deviation

lym_mean = malignant_pure_cores_counts['lym'].mean()
lym_med = malignant_pure_cores_counts['lym'].median()
lym_std = malignant_pure_cores_counts['lym'].std()
lym_mad = median_abs_deviation(malignant_pure_cores_counts['lym'], scale='normal')

def z_score(x, x_mean, x_std):
    return (x - x_mean) / x_std

def robust_z_score(x, x_median, x_mad):
    return (x - x_median) / x_mad

malignant_pure_cores_counts['robust_z_score'] = malignant_pure_cores_counts['lym'].apply(lambda x: robust_z_score(x, lym_med, lym_mad))

fig, ax = plt.subplots(figsize=(6,4))
sns.kdeplot(malignant_pure_cores_counts, x='robust_z_score', ax=ax)
```

```python
from scipy.stats import kruskal

fig, axs = plt.subplots(nrows=1, ncols=3, figsize=(12,5))
axs = axs.flatten()

for i, ratio in enumerate(['CD8', 'CD4', 'CD68']):
    sns.boxplot(sc_densities, x='cons_supercluster', y=ratio, ax=axs[i])
    # sns.swarmplot(sc_densities, x='cons_supercluster', y=ratio, ax=axs[i])
    # neg_class = sc_densities[sc_densities['risk_group'] == 'Low risk'][ratio]
    # pos_class = sc_densities[sc_densities['risk_group'] == 'High risk'][ratio]
    statistic, p_val = kruskal(
        sc_densities[sc_densities['cons_supercluster'] == 'Cold discohesive'][ratio].values,
        sc_densities[sc_densities['cons_supercluster'] == 'Cold cohesive'][ratio].values,
        sc_densities[sc_densities['cons_supercluster'] == 'Hot discohesive\n(low risk)'][ratio].values,
        sc_densities[sc_densities['cons_supercluster'] == 'Hot cohesive'][ratio].values,
        sc_densities[sc_densities['cons_supercluster'] == 'Hot discohesive\n(high risk)'][ratio].values)
    axs[i].hlines(sc_densities[ratio].max() + 0.3, xmin=0, xmax=4, color='black')
    if p_val < 0.001:
        axs[i].text(s=f'p < 0.001', y=sc_densities[ratio].max() + 0.4, x=.9, size=16)
    else:
        axs[i].text(s=f'p = {p_val:.3f}', y=sc_densities[ratio].max() + 0.4, x=.9, size=16)
    axs[i].set_xlabel(None)
    # y_lab = f'{ratio.replace("_", ":")}'
    axs[i].set_ylabel(f'{ratio} density')
    # axs[i].set_ylabel(rf'{ratio} $log_{{10}}$ density')
    axs[i].set_ylim([sc_densities[ratio].min() - 0.1, sc_densities[ratio].max() + 0.65])
    axs[i].set_xticklabels(['CD', 'CC', 'HD-LR', 'HC', 'HD-HR'], rotation=90)

fig.supxlabel('Supercluster', y=.06)

plt.tight_layout()
plt.savefig(os.path.join(main_path, 'analysis/multiplex-if/6-plex/mac_t_cell_counts_hot_discohesive.pdf'), dpi=300, bbox_inches='tight', pad_inches=0.1)
plt.show()
```
